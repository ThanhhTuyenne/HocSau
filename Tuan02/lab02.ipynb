```python
import torch

x8 = torch.tensor(3.6, requires_grad=True)
y8 = x8 * x8
y8.backward()
print("x8.grad (y = x² tại x=3.6):", x8.grad.item())
print("Đạo hàm thủ công 2x =", 2 * 3.6)

# Nâng cao: y = 3x² + 2x + 1
x8_2 = torch.tensor(3.6, requires_grad=True)
y8_2 = 3 * x8_2**2 + 2 * x8_2 + 1
y8_2.backward()
print("x8_2.grad (y = 3x² + 2x + 1):", x8_2.grad.item())
print("Đạo hàm thủ công 6x + 2 =", 6*3.6 + 2)
```

```python
import torch
import numpy as np

np.random.seed(42)
torch.manual_seed(42)

# Đa thức bậc 2: 2x² + 3x + 1
polynomial9 = np.poly1d([2, 3, 1])

N9 = 20
X9 = np.random.randn(N9, 1) * 5
Y9 = polynomial9(X9)

XX9 = np.hstack([X9 * X9, X9, np.ones_like(X9)])

w9 = torch.randn(3, 1, requires_grad=True)
x9 = torch.tensor(XX9, dtype=torch.float32)
y9 = torch.tensor(Y9, dtype=torch.float32)

optimizer9 = torch.optim.NAdam([w9], lr=0.01)

for epoch in range(1000):
    optimizer9.zero_grad()
    y_pred9 = x9 @ w9
    mse9 = torch.mean(torch.square(y9 - y_pred9))
    mse9.backward()
    optimizer9.step()

print("Learned coefficients:\n", w9.detach().numpy().flatten())
```

```python
import torch
import random

random.seed(42)
torch.manual_seed(42)
A10 = torch.tensor(random.random(), requires_grad=True)
B10 = torch.tensor(random.random(), requires_grad=True)
C10 = torch.tensor(random.random(), requires_grad=True)
D10 = torch.tensor(random.random(), requires_grad=True)
optimizer10 = torch.optim.NAdam([A10, B10, C10, D10], lr=0.01)

for _ in range(2000):
    y1 = A10 + B10 - 9
    y2 = C10 - D10 - 1
    y3 = A10 + C10 - 8
    y4 = B10 - D10 - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer10.zero_grad()
    sqerr.backward()
    optimizer10.step()

print(f"A={A10.item():.4f}, B={B10.item():.4f}, C={C10.item():.4f}, D={D10.item():.4f}")
```

```python
import torch
import numpy as np

# Ý 1
X11 = torch.arange(24, dtype=torch.float32)
print("Reshape 4x6 shape:", X11.reshape(4, 6).shape)
print("Reshape 2x12 shape:", X11.reshape(2, 12).shape)
print("Reshape 2x3x4 shape:", X11.reshape(2, 3, 4).shape)

# Ý 2
A11 = torch.randn(3, 4)
B11 = torch.randn(3, 4) + 1e-4
print("A+B:\n", A11 + B11)
print("A*B:\n", A11 * B11)
print("A/B:\n", A11 / B11)
print("Mean theo cột A:\n", torch.mean(A11, dim=0))
print("Std theo cột A:\n", torch.std(A11, dim=0))

# Ý 3
x11_3 = torch.tensor(2.0, requires_grad=True)
y11_3 = 2 * x11_3**2 + 5 * x11_3 + 3
y11_3.backward()
print("Autograd dy/dx tại x=2:", x11_3.grad.item())
print("Thủ công 4x+5 tại x=2:", 4*2 + 5)

# Ý 4
np.random.seed(42)
polynomial11_4 = np.poly1d([2, -3, 5])
N11 = 20
X11_arr = np.random.randn(N11, 1) * 5
Y11_arr = polynomial11_4(X11_arr)
XX11_arr = np.hstack([X11_arr * X11_arr, X11_arr, np.ones_like(X11_arr)])
x11_tensor = torch.tensor(XX11_arr, dtype=torch.float32)
y11_tensor = torch.tensor(Y11_arr, dtype=torch.float32)
w11_4 = torch.randn(3, 1, requires_grad=True)
optimizer11_4 = torch.optim.NAdam([w11_4], lr=0.01)

for _ in range(1000):
    optimizer11_4.zero_grad()
    y_pred11_4 = x11_tensor @ w11_4
    mse11_4 = torch.mean(torch.square(y11_tensor - y_pred11_4))
    mse11_4.backward()
    optimizer11_4.step()
print("Hệ số học được cho y = 2x² - 3x + 5:\n", w11_4.detach().numpy().flatten())
```
